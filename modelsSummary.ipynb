```python
from keras.models import load_model
```

```python
base_cnn = load_model('models/BaseCNN_mri_tumor.keras')
```

```python
base_cnn.summary()
```

```python
effnet = load_model('models/efficientNetB0_mri_tumor.keras')
```

```python
effnet.summary()
```

```python
resnet = load_model('models/resnet50_mri_tumor.keras')
```

```python
resnet.summary()
```

```python
models = [base_cnn, effnet, resnet]
for model in models:
    print(f"Model: {model.name}")
    print(f"Number of layers: {len(model.layers)}")
    print(f"Input shape: {model.input_shape}")
    print(f"Output shape: {model.output_shape}")
    print(f"Number of parameters: {model.count_params()}")
    for layer in model.layers:
        print(layer.name, layer.__class__.__name__)
    print("-" * 50)
```
